```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
```

```python
df = pd.read_csv("data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.select_dtypes(include=['int64','float64']).columns
```

```python
print("Target Variable : diagnosis")
```

```python
df.isnull().sum()
```

```python
 df.drop(['id','Unnamed: 32'], axis=1, inplace=True)
```

```python
le = LabelEncoder()

df['diagnosis'] = le.fit_transform(df['diagnosis'])

df.head()
```

```python
X = df.drop('diagnosis', axis=1)

y = df['diagnosis']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
      X,
          y,
              test_size=0.20,
                  random_state=42
                  )
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
model = KNeighborsClassifier(n_neighbors=5)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

y_pred
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
```

```python
precision = precision_score(y_test, y_pred)

print("Precision :", precision)
```

```python
recall = recall_score(y_test, y_pred)

print("Recall :", recall)
```

```python
f1 = f1_score(y_test, y_pred)

print("F1 Score :", f1)
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot()

plt.title("Confusion Matrix")

plt.show()
```

## Observations

1. The KNN model achieved good classification accuracy.

2. Feature scaling improved the model because KNN uses distance calculations.

3. The confusion matrix shows that most tumors were classified correctly.

## Conclusion

The K-Nearest Neighbors (KNN) classifier was successfully used to classify breast tumors as Malignant or Benign. The dataset was preprocessed by removing unnecessary columns, encoding the target variable, and standardizing the feature values before training the model with K = 5. The model was evaluated using Accuracy, Precision, Recall, F1-Score, and the Confusion Matrix. The results indicate that KNN performs well for this classification task. Feature scaling is essential because KNN is a distance-based algorithm. One limitation of KNN is that it becomes slower for large datasets since it computes distances to all training samples during prediction.
